# 3. Negativity tests

Remember that the negativity of a normal (diagonalizable) operator is defined as follows:

>\begin{equation}
    n(\rho) = \sum _{i:\lambda _i <0} |\lambda _i|
\end{equation}

Where $\lambda _i$ are the eigenvalues of the partial transpose of $\rho$.

## 3.3. Some observations

### 3.3.1.

These files contain separable states. Their partial transpose is again positive, so their negativity must be zero. However, when we compute their negativity, we get values of order 10^-6 and 10^-7. I believe these are numerical errors because, for example, the unit-trace test fails for tolerances of 10^-6 or smaller. The tolerance has to be raised to 10^-5 for the test to pass on all matrices. So it should not surprise me that both the unit trace and the zero negativity are exact only up to order O(10^-5).

### 3.3.2.

Note that the partial transpose $\rho^{T_B}$ is defined for every density matrix, separable or entangled: in the computational basis, transposing the second qubit swaps the matrix elements $\langle ij|\rho|kl\rangle \to \langle il|\rho|kj\rangle$, and transposing the first qubit swaps $\langle ij|\rho|kl\rangle \to \langle kj|\rho|il\rangle$. This is exactly what our element swapping does, and both choices give the same eigenvalues. It always preserves Hermiticity: if $\rho=\rho^\dagger$, then $(\rho^{T_B})^\dagger=\rho^{T_B}$. What it does **not** preserve in general is positivity. For a separable state, $\rho^{T_B}$ is again a valid density matrix, with no negative eigenvalues. For an entangled two-qubit state, $\rho^{T_B}$ always has a negative eigenvalue (Peres–Horodecki criterion), and the negativity measures it. Since $\rho^{T_B}$ is always Hermitian, we can use np.linalg.eigh to compute its eigenvalues and, from them, the negativity.
